## Anylyzing Data from data frame

Q . What is the total number of reported cases and deaths related to covid-19 in Italy ? 

Similar to NumPy arrays , Pandas series supports the sum  method to answer these questions.

```python
total_cases = covid_df.new_cases.sum()
total_deaths = covid_df.new_deaths.sum()
```

```python
print(f'The number of reported cases in {total_cases} and the number of deaths is {total_deaths} ')
```

Q : What is THe overall deaths rate (ratio of reported deats to reported cases)?


```python
death_rate = covid_df.new_deaths.sum()/covid_df.new_cases.sum()
```

```python
print("THe overall reported death rate in italy is {:.2f}%.".format(death_rate))
```

Q : What is the overall number of tests conducted ? A total of 935310 tests were conducted before daily test numbers were being reported .


we can check the first non-NaN index using first_valid_index

```python
initial_tests = 935310

total_tests = initial_tests + covid_df.new_tests.sum()
```

```python
total_tests
```

Q : What fraction of test returned a positive result ?
 

```python
positive_rate = total_cases / total_tests
```

```python
print('{:.2f}%  of tests in Italy led to a positive diagnosis.'.format(positive_rate))
```

## Querying and Sorting rows 

let's say we want only want to look at the days which had more than 1000 reported cases . We can use a boolean expression to check which rows satisfy this criterion. 

```python
high_new_cases = covid_df.new_cases > 1000
```

```python
high_new_cases 
```

```python
high_new_cases 
```

The boolean expression returns a series containing True and False boolean values. This series can be used to filter our just those rows where the value in the series is True . The result is data frame with a subset of rows from the original.

```python
covid_df[high_new_cases]
```

The data frame contains 72 rows , but only the first 5 and last 5 rows are displayed by default with Jupyter , for brevity . To views , all the rows , we can modify some display options .

```python
from IPython.display import display 
with pd.option_context('display.max_rows',100):
    display(covid_df[covid_df.new_cases > 1000])
```

we can also formulate more complex queries that involve multiple columns . As an example , let's try to determine the days when the ratio of cases of cases reported to tests conducted  is higher than the overall positive_rate.

```python
positive_rate
```

```python
high_ratio_df = covid_df[covid_df.new_cases / covid_df.new_tests > positive_rate]
```

```python
high_ratio_df
```

Performing operations on multiple columns results in a new series 

```python
covid_df.new_cases / covid_df.new_tests
```

Further , we can use this series to add a new column to the data frame.

```python
covid_df['positive_rate'] = covid_df.new_cases / covid_df.new_tests
```
